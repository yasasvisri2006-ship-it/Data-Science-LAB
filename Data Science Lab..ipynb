```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

df = pd.read_csv(url)

print("First Five Records:")
print(df.head())

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())

plt.figure(figsize=(8, 5))

plt.plot(df.index, df["sepal_length"], label="Sepal Length")
plt.plot(df.index, df["petal_length"], label="Petal Length")

plt.xlabel("Sample Index")
plt.ylabel("Length")
plt.title("Sepal Length and Petal Length")
plt.legend()
plt.grid()
plt.show()

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="sepal_length",
    y="petal_length",
    hue="species"
)

plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Petal Length")
plt.show()

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="sepal_length",
    hue="species",
    kde=True
)

plt.title("Distribution of Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="species",
    y="petal_length"
)

plt.title("Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length")
plt.show()

sns.pairplot(
    df,
    hue="species"
)

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [120, 150, 180, 160, 220, 250]
expenses = [80, 100, 120, 110, 140, 160]

fig, ax = plt.subplots(2, 1, figsize=(10, 8))

ax[0].plot(
    months,
    sales,
    marker="o",
    linewidth=2,
    label="Sales"
)

ax[0].set_title("Monthly Sales")
ax[0].set_xlabel("Month")
ax[0].set_ylabel("Sales")

ax[0].set_xticks(range(len(months)))
ax[0].set_xticklabels(months)

ax[0].annotate(
    "Highest Sales",
    xy=(5, 250),
    xytext=(3.5, 270),
    arrowprops=dict(arrowstyle="->")
)

ax[0].legend()
ax[0].grid(True)

ax[1].plot(
    months,
    expenses,
    marker="s",
    linewidth=2,
    label="Expenses"
)

ax[1].set_title("Monthly Expenses")
ax[1].set_xlabel("Month")
ax[1].set_ylabel("Expenses")

ax[1].set_xticks(range(len(months)))
ax[1].set_xticklabels(months)

ax[1].annotate(
    "Highest Expense",
    xy=(5, 160),
    xytext=(3.5, 180),
    arrowprops=dict(arrowstyle="->")
)

ax[1].legend()
ax[1].grid(True)

fig.suptitle("Monthly Sales and Expenses Analysis", fontsize=16)

plt.tight_layout()

plt.savefig(
    "monthly_sales_expenses.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import pandas as pd

india_dates = pd.date_range(
    start='2026-01-01 09:00',
    periods=3,
    freq='h',
    tz='Asia/Kolkata'
)

print("1. Date Range with Asia/Kolkata Time Zone:")
print(india_dates)

dates = pd.date_range(
    start='2026-01-01 09:00',
    periods=3,
    freq='h'
)

print("\n2. Timezone-naive Date Range:")
print(dates)

localized_dates = dates.tz_localize('Asia/Kolkata')

print("\n3. After Localizing to Asia/Kolkata:")
print(localized_dates)

new_york_dates = localized_dates.tz_convert(
    'America/New_York'
)

print("\n4. Converted to America/New_York:")
print(new_york_dates)

india_series = pd.Series(
    [100, 200, 300],
    index=localized_dates
)

new_york_series = pd.Series(
    [400, 500, 600],
    index=new_york_dates
)

print("\n5. India Time Series:")
print(india_series)

print("\nNew York Time Series:")
print(new_york_series)

combined_series = pd.concat([
    india_series,
    new_york_series
])

print("\n6. Combined Time Series:")
print(combined_series)
```

```python
import pandas as pd

p = pd.Period('2026-01', freq='M')

print("Original Period:")
print(p)

daily_start = p.asfreq('D', how='start')

print("\nMonthly Period converted to Daily (Start):")
print(daily_start)

daily_end = p.asfreq('D', how='end')

print("\nMonthly Period converted to Daily (End):")
print(daily_end)

period_index = pd.period_range(
    start='2026-01',
    periods=3,
    freq='M'
)

print("\nOriginal PeriodIndex:")
print(period_index)

daily_period_index = period_index.asfreq(
    'D',
    how='start'
)

print("\nPeriodIndex converted to Daily Frequency:")
print(daily_period_index)

daily_period_index_end = period_index.asfreq(
    'D',
    how='end'
)

print("\nPeriodIndex converted to Daily Frequency (End):")
print(daily_period_index_end)
```

```python
import pandas as pd

dates = pd.date_range(
    start='2026-01-01 00:00',
    periods=12,
    freq='h'
)

values = [10, 12, 15, 14, 18, 20,
          22, 21, 25, 28, 30, 32]

time_series = pd.Series(
    values,
    index=dates
)

print("Original Hourly Time Series:")
print(time_series)

resampled = time_series.resample('3h').mean()

print("\nResampled Time Series - 3 Hour Mean:")
print(resampled)

downsampled = time_series.resample('4h').sum()

print("\nDownsampled Time Series - 4 Hour Sum:")
print(downsampled)

upsampled = time_series.resample('30min').asfreq()

print("\nUpsampled Time Series - 30 Minute:")
print(upsampled)

upsampled_ffill = time_series.resample('30min').ffill()

print("\nUpsampled Time Series using Forward Fill:")
print(upsampled_ffill)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

df = pd.read_csv("product_reviews.csv")
df["review"] = df["review"].fillna("").astype(str)

analyzer = SentimentIntensityAnalyzer()

def classify_sentiment(text):
    score = analyzer.polarity_scores(text)["compound"]
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    return "Neutral"

df["sentiment"] = df["review"].apply(classify_sentiment)

def generate_wordcloud(text, title, filename):
    if not text.strip():
        print(f"No text available for: {title}")
        return

    cloud = WordCloud(
        width=1200,
        height=600,
        background_color="white",
        max_words=150,
        collocations=False
    ).generate(text)

    plt.figure(figsize=(12, 6))
    plt.imshow(cloud, interpolation="bilinear")
    plt.axis("off")
    plt.title(title)
    plt.tight_layout()
    plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()

all_text = " ".join(df["review"])

generate_wordcloud(
    all_text,
    "WordCloud - All Reviews",
    "wordcloud_all.png"
)

positive_text = " ".join(
    df.loc[df["sentiment"] == "Positive", "review"]
)

generate_wordcloud(
    positive_text,
    "WordCloud - Positive Reviews",
    "wordcloud_positive.png"
)

negative_text = " ".join(
    df.loc[df["sentiment"] == "Negative", "review"]
)

generate_wordcloud(
    negative_text,
    "WordCloud - Negative Reviews",
    "wordcloud_negative.png"
)

df.to_csv("product_reviews_analyzed.csv", index=False)

print(df["sentiment"].value_counts())
print("WordCloud images saved successfully.")
```

```python
import re
import pandas as pd
import nltk
import matplotlib.pyplot as plt
import seaborn as sns
from nltk.corpus import stopwords
from collections import Counter
from tkinter import Tk, filedialog

nltk.download("stopwords", quiet=True)

root = Tk()
root.withdraw()

file_path = filedialog.askopenfilename(
    title="Select product_reviews.csv",
    filetypes=[("CSV files", "*.csv"), ("All files", "*.*")]
)

if not file_path:
    print("No file selected.")
else:
    df = pd.read_csv(file_path)

    if "review" not in df.columns:
        print("Error: 'review' column not found.")
        print("Available columns:", list(df.columns))
    else:
        df["review"] = df["review"].fillna("").astype(str)

        stop_words = set(stopwords.words("english"))

        def analyze_review(text):
            tokens = re.findall(r"\b[\w']+\b", text.lower())
            words = re.findall(r"\b[A-Za-z]+\b", text)
            numeric_tokens = re.findall(r"\b\d+(?:\.\d+)?\b", text)

            stopword_tokens = [
                token for token in tokens if token in stop_words
            ]

            return pd.Series({
                "word_count": len(tokens),
                "character_count": len(text),
                "numeric_count": len(numeric_tokens),
                "stopword_count": len(stopword_tokens),
                "average_word_length": (
                    sum(len(word) for word in words) / len(words)
                    if words else 0
                )
            })

        statistics = df["review"].apply(analyze_review)

        df = pd.concat([df, statistics], axis=1)

        df.to_csv("review_statistics.csv", index=False)

        columns = [
            ("stopword_count", "Stopword Count"),
            ("numeric_count", "Numeric Token Count"),
            ("word_count", "Word Count"),
            ("character_count", "Character Count"),
            ("average_word_length", "Average Word Length")
        ]

        fig, axes = plt.subplots(3, 2, figsize=(14, 15))
        axes = axes.flatten()

        for i, (column, title) in enumerate(columns):
            sns.histplot(
                data=df,
                x=column,
                kde=True,
                ax=axes[i],
                bins=20
            )

            axes[i].set_title(title)
            axes[i].set_xlabel(title)
            axes[i].set_ylabel("Number of Reviews")

        axes[-1].axis("off")

        plt.tight_layout()
        plt.savefig(
            "review_statistics_distributions.png",
            dpi=300
        )
        plt.show()

        all_tokens = re.findall(
            r"\b[\w']+\b",
            " ".join(df["review"]).lower()
        )

        stopword_frequency = Counter(
            token for token in all_tokens
            if token in stop_words
        )

        print("Top 10 stopwords:")
        print(stopword_frequency.most_common(10))

        print("\nDescriptive statistics:")
        print(
            df[
                [
                    "stopword_count",
                    "numeric_count",
                    "word_count",
                    "character_count",
                    "average_word_length"
                ]
            ].describe()
        )

        print("\nAnalysis completed successfully.")
        print("Saved: review_statistics.csv")
        print("Saved: review_statistics_distributions.png")
```

```python
import pandas as pd
import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer as NLTKVADER
from vaderSentiment.vaderSentiment import (
    SentimentIntensityAnalyzer as StandaloneVADER
)

nltk.download("vader_lexicon", quiet=True)

df = pd.read_csv("product_reviews.csv")
df["review"] = df["review"].fillna("").astype(str)

nltk_analyzer = NLTKVADER()
vader_analyzer = StandaloneVADER()

def classify(score):
    if score >= 0.05:
        return "Positive"
    elif score <= -0.05:
        return "Negative"
    return "Neutral"

nltk_scores = df["review"].apply(
    nltk_analyzer.polarity_scores
)

df["nltk_positive"] = nltk_scores.apply(lambda x: x["pos"])
df["nltk_negative"] = nltk_scores.apply(lambda x: x["neg"])
df["nltk_neutral"] = nltk_scores.apply(lambda x: x["neu"])
df["nltk_compound"] = nltk_scores.apply(lambda x: x["compound"])

vader_scores = df["review"].apply(
    vader_analyzer.polarity_scores
)

df["vader_positive"] = vader_scores.apply(lambda x: x["pos"])
df["vader_negative"] = vader_scores.apply(lambda x: x["neg"])
df["vader_neutral"] = vader_scores.apply(lambda x: x["neu"])
df["vader_compound"] = vader_scores.apply(lambda x: x["compound"])

df["nltk_sentiment"] = df["nltk_compound"].apply(classify)
df["vader_sentiment"] = df["vader_compound"].apply(classify)

print(df[[
    "review",
    "nltk_compound",
    "nltk_sentiment",
    "vader_compound",
    "vader_sentiment"
]].head(10).to_string(index=False))

print("\nNLTK sentiment distribution:")
print(df["nltk_sentiment"].value_counts())

print("\nStandalone VADER sentiment distribution:")
print(df["vader_sentiment"].value_counts())

df.to_csv("sentiment_analysis_results.csv", index=False)

print("\nResults saved to sentiment_analysis_results.csv")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from tkinter import Tk, filedialog

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

root = Tk()
root.withdraw()

file_path = filedialog.askopenfilename(
    title="Select marketing_campaign.csv",
    filetypes=[("CSV files", "*.csv"), ("All files", "*.*")]
)

if not file_path:
    print("No file selected.")
else:
    df = pd.read_csv(file_path, sep="\t")

    print("Dataset loaded successfully.")
    print("Rows:", len(df))
    print("Columns:", len(df.columns))

    features = [
        "Income",
        "MntWines",
        "MntFruits",
        "MntMeatProducts",
        "MntFishProducts",
        "MntSweetProducts",
        "MntGoldProds"
    ]

    missing_columns = [
        column for column in features
        if column not in df.columns
    ]

    if missing_columns:
        print("Missing columns:", missing_columns)
        print("Available columns:")
        print(df.columns.tolist())

    else:
        data = df[features].copy()

        data = data.fillna(data.median())

        data = data.clip(
            lower=data.quantile(0.01),
            upper=data.quantile(0.99),
            axis=1
        )

        scaler = StandardScaler()

        X = scaler.fit_transform(data)

        inertias = []
        scores = []

        for k in range(2, 8):
            model = KMeans(
                n_clusters=k,
                random_state=42,
                n_init=10
            )

            labels = model.fit_predict(X)

            inertias.append(model.inertia_)
            scores.append(silhouette_score(X, labels))

        plt.figure(figsize=(8, 5))

        plt.plot(
            range(2, 8),
            inertias,
            marker="o"
        )

        plt.xlabel("Number of Clusters (K)")
        plt.ylabel("Inertia")
        plt.title("Elbow Method")
        plt.grid(True)
        plt.show()

        print(
            "Silhouette scores:",
            dict(zip(range(2, 8), scores))
        )

        best_k = range(2, 8)[
            scores.index(max(scores))
        ]

        print("Best number of clusters:", best_k)

        kmeans = KMeans(
            n_clusters=best_k,
            random_state=42,
            n_init=10
        )

        df["Customer_Segment"] = kmeans.fit_predict(X)

        summary = (
            df.groupby("Customer_Segment")[features]
            .mean()
            .round(2)
        )

        print("\nCustomer Segment Summary:")
        print(summary)

        plt.figure(figsize=(10, 6))

        sns.scatterplot(
            data=df,
            x="Income",
            y="MntWines",
            hue="Customer_Segment",
            palette="Set2"
        )

        plt.title(
            "Customer Segments by Income and Wine Spending"
        )

        plt.xlabel("Income")
        plt.ylabel("Wine Spending")

        plt.show()

        df.to_csv(
            "customer_segments.csv",
            index=False
        )

        summary.to_csv(
            "segment_summary.csv"
        )

        print("\nAnalysis completed successfully.")
        print("Saved: customer_segments.csv")
        print("Saved: segment_summary.csv")
```

```python
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

data = {
    "text": [
        "I am very happy today",
        "This is a wonderful day",
        "I feel cheerful and excited",
        "I am sad and lonely",
        "I feel very disappointed",
        "I miss my family so much",
        "I am angry with you",
        "This makes me furious",
        "I hate this situation",
        "I am scared of the dark",
        "I am afraid of losing",
        "I feel nervous and frightened",
        "I love you so much",
        "You mean everything to me",
        "I adore spending time with you",
        "Wow this is unbelievable",
        "I did not expect this surprise",
        "What an amazing unexpected gift"
    ],
    "emotion": [
        "happy", "happy", "happy",
        "sadness", "sadness", "sadness",
        "anger", "anger", "anger",
        "fear", "fear", "fear",
        "love", "love", "love",
        "surprise", "surprise", "surprise"
    ]
}

df = pd.DataFrame(data)

X_train, X_test, y_train, y_test = train_test_split(
    df["text"],
    df["emotion"],
    test_size=0.33,
    random_state=42,
    stratify=df["emotion"]
)

model = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2))),
    ("classifier", LogisticRegression(max_iter=1000))
])

model.fit(X_train, y_train)

predictions = model.predict(X_test)

print(
    "Test Accuracy:",
    round(accuracy_score(y_test, predictions) * 100, 2),
    "%"
)

print("\nClassification Report:")

print(
    classification_report(
        y_test,
        predictions,
        zero_division=0
    )
)

emoji_map = {
    "happy": "😊",
    "sadness": "😢",
    "anger": "😠",
    "fear": "😨",
    "love": "❤️",
    "surprise": "😲"
}

def detect_emotion(text):
    emotion = model.predict([text])[0]
    emoji = emoji_map[emotion]

    print("Input Text:", text)
    print("Detected Emotion:", emotion)
    print("Relevant Emoji:", emoji)

detect_emotion("I am happy to see you")
detect_emotion("I feel lonely today")
detect_emotion("I love my family")
```
