```python
import pandas as pd
student_data = {
    "Roll No": [101, 102, 103, 104],
    "Name": ["Yasasvi", "Charitha", "Reshma", "Preethi"],
    "Department": ["CSE(DS)", "CSE(DS)", "IT", "CSE"],
    "Percentage": [89, 92, 88, 85]
}
course_data = {
    "course_ID": [101,102,103,104],
    "Course_Name" : ["Python","Data Science","Machine Learning","Java"],
    "Credits":[4,3,4,3]
}
Student_df = pd.DataFrame(student_data)
course_df = pd.DataFrame(course_data)
print("Student Data:")
print(Student_df)
print("\nCourse Data:")
print(course_df)
```

```python
#Writing multiple Dataframes to different sheets 
with pd.ExcelWriter("college-data.xlsx",engine="openpyxl") as writer:
    Student_df.to_excel(
        writer,
        sheet_name="Students",
        index=False
    )

    course_df.to_excel(
        writer,
        sheet_name="Courses",
        index=False
    )

print("Multiple sheets successfully written to college-data.xlsx")
    
     
    
```

```python
#complete combined program
import pandas as pd
from io import StringIO
#---------------------
# 1.CSV READING AND WRITING
#----------------------------
csv_url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
csv_df = pd.read_csv(csv_url)
print("CSV Data:")
print(csv_df.head())
csv_df.to_csv("iris_output.csv",index=False)
print("\nCSV file written successfully")
#---------------------------------------
#2.TAB-DELIMITED FILE
#--------------------------------
student_data = {
    "Roll_No" : [101,102,103,104],
    "Name": ["Yasasvi","Charitha","Reshma","Preethi"],
    "Department": ["CSE(DS)", "CSE(DS)", "IT", "CSE"],
    "Percentage": [89, 92, 88, 85]
}
Student_df = pd.DataFrame(student_data)
Student_df.to_csv(
    "Student_output.txt",
    sep="\t",
    index=False
)
tab_df = pd.read_csv(
    "Student_output.txt",
    sep="\t"
)
print("\nTab-Delimited Data:")
print(tab_df)
#------------------------------------
#3.Json Reading and writing
#------------------------------------
Student_df.to_json(
    "Student_output.json",
    indent=4
)
json_df = pd.read_json("Student_output.json")
print("\nJSON Data:")
print(json_df)
    
```

```python
import sqlite3
import pandas as pd
# connect to database
conn = sqlite3.connect("Students.db")
cursor = conn.cursor()
# create table
cursor.execute("""
CREATE TABLE IF NOT EXISTS STUDENTS (
ID INTEGER,
NAME TEXT
)
""")
# insert table
cursor.execute("INSERT INTO Students VALUES(1,'john')")
conn.commit()
cursor.execute("INSERT INTO Students VALUES(2,'joy')")
conn.commit()
# Read data
df = pd.read_sql_query("SELECT * FROM Students",conn)
print("Student Data:")
print(df)
# update and delete
cursor.execute("UPDATE Students SET NAME = 'Alice' WHERE ID = 1")
cursor.execute("DELETE FROM Students WHERE ID = 2")
conn.commit()
# Read updated data
df = pd.read_sql_query("SELECT * FROM Students", conn)
print("\nUpdated Student Data:")
print(df)
conn.close()
```
